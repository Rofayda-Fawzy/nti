```python
#ex 2
prices=map(float,input("enter 3 numbers of product prices in on line").split())
for price in prices :
    if price <=50 :
        print(price,"is cheap")
    elif price <=100 :
        print(price,"is moderate")
    else  :
        print(price,"is expensive")
```

```python
# ex 3
inp=input("enter word with even-length")
inp=inp[::2]
text=list(inp)
last=text[-1]
last2=text[-2]
text=inp[:-2]+last+last2
text=text[-1]+text[1:-1]+text[0]

print(text)
```

```python
#ex 3 anthor solution
inp=input("enter word with even-length")
inp=inp[::2]
mid=len(inp)//2
text1=list(inp[:mid])
text2=list(inp[mid:])
text2_rev=text2[::-1]
text3=''.join(text1+text2_rev)
final_text=(text3[-1]+text3[1:-1]+text3[0]).upper()
print(final_text)
```

```python
# ex 4
matrix=[[0 for i in range(3)] for j in range(3)]
for i in range(3):
    for j in range(3):
        if i==1 and i+j== 2:
            matrix[i][j]= 3
        elif i==j:
            matrix[i][j]= 1
        elif i+j==2:
            matrix[i][j]= 2
       

for row in matrix:
    print(row)
```
